```python
from pathlib import Path

import pandas as pd
import numpy as np
```

```python
PROJECT_ROOT = Path.cwd().parent

DATA_DIR = PROJECT_ROOT / "data" / "raw"

PROJECT_ROOT, DATA_DIR
```

```python
brfss_2015_path = (
    DATA_DIR
    / "brfss_2015"
    / "diabetes_binary_health_indicators_BRFSS2015.csv"
)

brfss_2021_path = (
    DATA_DIR
    / "brfss_2021"
    / "diabetes_binary_health_indicators_BRFSS2021.csv"
)

clinical_diabetes_path = (
    DATA_DIR
    / "clinical_diabetes"
    / "diabetes_prediction_dataset.csv"
)
```

```python
print("BRFSS 2015 exists:", brfss_2015_path.exists())
print("BRFSS 2021 exists:", brfss_2021_path.exists())
print("Clinical dataset exists:", clinical_diabetes_path.exists())
```

```python
brfss_2015 = pd.read_csv(brfss_2015_path)
brfss_2021 = pd.read_csv(brfss_2021_path)
clinical_diabetes = pd.read_csv(clinical_diabetes_path)
```

```python
def inspect_dataset(df, name, target_column=None):
    print("=" * 80)
    print(f"DATASET: {name}")
    print("=" * 80)

    print(f"\nShape: {df.shape}")

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData Types:")
    display(df.dtypes)

    print("\nFirst 5 Rows:")
    display(df.head())

    print("\nMissing Values:")
    missing = df.isnull().sum()
    display(missing[missing > 0])

    if missing.sum() == 0:
        print("No missing values.")

    print(f"\nDuplicate Rows: {df.duplicated().sum()}")

    print("\nUnique Values:")
    display(df.nunique().sort_values())

    if target_column and target_column in df.columns:
        print(f"\nTarget Distribution: {target_column}")

        target_counts = df[target_column].value_counts()
        target_percentage = (
            df[target_column]
            .value_counts(normalize=True)
            .mul(100)
            .round(2)
        )

        target_summary = pd.DataFrame({
            "Count": target_counts,
            "Percentage": target_percentage
        })

        display(target_summary)

    print("\nBasic Statistics:")
    display(df.describe().T)
```

```python
inspect_dataset(
    brfss_2015,
    "BRFSS 2015 - Binary Diabetes",
    target_column="Diabetes_binary"
)
```

```python
inspect_dataset(
    brfss_2021,
    "BRFSS 2021 - Binary Diabetes",
    target_column="Diabetes_binary"
)
```

```python
inspect_dataset(
    clinical_diabetes,
    "Clinical Diabetes Prediction Dataset",
    target_column="diabetes"
)
```

```python
brfss_2015_columns = set(brfss_2015.columns)
brfss_2021_columns = set(brfss_2021.columns)
clinical_columns = set(clinical_diabetes.columns)

print("Common columns between BRFSS 2015 and 2021:")
print(sorted(brfss_2015_columns & brfss_2021_columns))

print("\nOnly in BRFSS 2015:")
print(sorted(brfss_2015_columns - brfss_2021_columns))

print("\nOnly in BRFSS 2021:")
print(sorted(brfss_2021_columns - brfss_2015_columns))

print("\nCommon columns between BRFSS 2015 and Clinical dataset:")
print(sorted(brfss_2015_columns & clinical_columns))
```
